### Importancia Industrial: Ectoína e Hidroxiectoína en Dermocosmética

Para sobrevivir a presiones osmóticas letales sin acumular concentraciones intracelulares tóxicas de iones inorgánicos (la estrategia "*salt-in*"), *Halomonas elongata* emplea la estrategia de **solutos compatibles** (la estrategia "*salt-out*"). Sintetiza y acumula en su citoplasma osmolitos zwitteriónicos de bajo peso molecular, principalmente **ectoína** (ácido 1,4,5,6-tetrahidro-2-metil-4-pirimidincarboxílico) e **hidroxiectoína**.

Estos compuestos poseen propiedades biofísicas extraordinarias:
1. **Estabilización macromolecular:** Forman nanoestructuras de hidratación protectoras ("*water clusters*") alrededor de proteínas, ácidos nucleicos y bicapas lipídicas, previniendo su desnaturalización térmica, agregación o daño oxidativo.
2. **Aplicaciones dermocosméticas y biomédicas:** Son ingredientes bioactivos de altísimo valor comercial en formulaciones antienvejecimiento, hidratación dérmica profunda, reparación de la barrera cutánea y protección contra radiación UV, luz azul y contaminación ambiental. En farmacia, se utilizan para estabilizar anticuerpos monoclonales, enzimas diagnósticas y vacunas.

### El Marcador Filogenético de Alta Resolución: Gen *gyrB* (GO:0003918)
Tradicionalmente, la taxonomía bacteriana se ha basado en el gen **16S ARNr**. Sin embargo, en el género *Halomonas* y en muchas bacterias halófilas, el 16S ARNr presenta una tasa de sustitución evolutiva muy lenta: especies congenéricas distintas frecuentemente comparten más del 98.5% al 99.5% de similitud de secuencia en el 16S ARNr, lo que imposibilita discriminar especies crípticas o resolver con precisión linajes filogenéticos estrechamente emparentados.

Para superar esta limitación, en este parcial emplearemos como marcador molecular de alta resolución el gen ***gyrB***:
* **Función biológica:** Codifica para la **subunidad B de la ADN girasa** (topoisomerasa tipo II bacteriana dependiente de ATP), asignada bajo el término de Ontología Génica **GO:0003918** (*DNA topoisomerase type II (ATP-hydrolyzing) activity*).
* **Propiedades genómicas:** Es una enzima esencial universal de copia única en bacterias que introduce superenrollamiento negativo en el ADN durante la replicación y transcripción.
* **Ventaja filogenética:** Al ser un gen codificante de proteína sometido a presión selectiva funcional pero con alta flexibilidad en las terceras posiciones de los codones (mutaciones sinónimas), exhibe una **tasa de sustitución nucleotídica significativamente mayor que el 16S ARNr**. Esto confiere un poder de resolución taxonómica superior para discriminar especies congenéricas y establecer filogenias robustas en la familia *Halomonadaceae*.

## Punto 1: Preprocesamiento y Control de Calidad de Lecturas Crudas

### Contexto Bioinformático
La tecnología de secuenciación masiva por síntesis de Illumina genera millones de lecturas cortas pareadas (*paired-end*). Cada base nitrogenada identificada está asociada a una puntuación de calidad **Phred ($Q$)**, definida de acuerdo con la probabilidad de error en el llamado de la base ($P$):

$$Q = -10 \log_{10}(P)$$

Un puntaje $Q \ge 20$ representa una precisión del $99\%$ (1 error cada 100 bases), mientras que $Q \ge 30$ indica una precisión del $99.9\%$ (1 error cada 1000 bases). 

Durante las corridas de secuenciación se presentan dos problemas sistemáticos habituales:
1. **Decaimiento de la calidad hacia el extremo 3':** Conforme avanzan los ciclos de secuenciación, el agotamiento gradual de reactivos y la desincronización de los clústeres (*dephasing*) reducen la fidelidad del fluoróforo, provocando un aumento en la tasa de error al final de las lecturas.
2. **Contaminación por adaptadores sintéticos:** Cuando el fragmento de ADN genómico secuenciado es más corto que el número de ciclos de secuenciación, el secuenciador continúa leyendo la secuencia del adaptador sintético ligado en el extremo 3' del inserto. La presencia de estos adaptadores distorsiona severamente los algoritmos de ensamblaje *de novo*.

## Punto 2: Ensamblaje *De Novo* y Evaluación Genómica

### Contexto Bioinformático
El ensamblaje genómico *de novo* consiste en reconstruir la secuencia completa del cromosoma bacteriano a partir de millones de lecturas cortas superpuestas, sin utilizar un genoma de referencia previo. 

* **Algoritmos basados en Grafos de De Bruijn (SPAdes):** Los secuenciadores de lecturas cortas dividen cada lectura en subcadenas de longitud fija llamadas *$k$-mers*. Los grafos de De Bruijn modelan las superposiciones de longitud $k-1$ entre estos $k$-mers como nodos y aristas. SPAdes utiliza una aproximación multi-$k$-mer adaptativa que combina valores pequeños de $k$ (para ensamblar regiones de baja cobertura) con valores grandes de $k$ (para resolver regiones repetitivas del genoma bacteriano). El modo `--isolate` optimiza el ensamblaje de bacterias aisladas con alta cobertura.
* **Métricas de Contigüidad Genómica (QUAST):** La contigüidad mide el grado de fragmentación del ensamblaje en **contigs** (secuencias contiguas continuas). La métrica reina es el **N50**: la longitud mínima del contig tal que la suma de las longitudes de todos los contigs de tamaño igual o superior cubre al menos el 50% de la longitud total ensamblada. A mayor N50 y menor número total de contigs, mayor es la continuidad del ensamblaje. Además, se evalúa el contenido de **%GC** (característico de cada grupo taxonómico).
* **Métricas de Completitud Biológica (BUSCO):** Un ensamblaje con alto N50 podría, no obstante, carecer de genes esenciales por sesgos de amplificación. BUSCO (*Benchmarking Universal Single-Copy Orthologs*) evalúa la presencia, fragmentación o pérdida de un conjunto predeterminado de genes de copia única evolutivamente universales y conservados en bacterias (`bacteria_odb10`).

## Punto 4: Genómica Comparativa y Detección de Organismos Fuera de Distribución (OFDs) con PCoA

### Contexto Bioinformático
Para evaluar el parentesco y ubicar evolutivamente nuestro aislado, debemos compararlo con la colección de secuencias homólogas de *gyrB* presentes en `./data/coleccion_gyrB.fasta`.

* **Alineamiento Múltiple de Secuencias (MUSCLE):** Alinea de forma progresiva e iterativa las secuencias nucleotídicas, insertando gaps para maximizar la identidad posicional entre sitios homólogos.
* **Matriz de Distancias de Hamming:** Sobre el alineamiento de secuencias de longitud $L$, la distancia de Hamming no corregida calcula la fracción de posiciones con nucleótidos distintos entre cada par de secuencias $i$ y $j$:
  $$D_H(i, j) = \frac{\text{Número de sitios disímiles}}{L}$$
* **Análisis de Coordenadas Principales (PCoA / MDS Clásico):** A diferencia del PCA estándar (que opera sobre matrices de características continuas), el PCoA (*Principal Coordinates Analysis*) o Escalamiento Multidimensional Métrico (*Metric Multidimensional Scaling*) toma directamente una **matriz de distancias pareadas** ($N \times N$) y proyecta las secuencias en un espacio ortogonal de baja dimensión (habitualmente 2D: coordenadas PC1 y PC2) preservando al máximo las distancias geométricas observadas.
* **Organismos Fuera de Distribución (OFDs):** En estudios metagenómicos o colecciones ambientales es común encontrar secuencias espurias, artefactos de PCR, quimeras de secuenciación o secuencias de taxones lejanos (p. ej., de otros filos no halófilos) que contaminan la muestra. Estos organismos actúan como **OFDs (Organismos Fuera de la Distribución)**: aparecen como puntos extremadamente aislados en el PCoA. Si no se remueven antes de la reconstrucción filogenética, distorsionan el alineamiento múltiple e introducen el grave artefacto de **atracción de ramas largas** (*Long Branch Attraction - LBA*). Por tanto, su detección y eliminación es un paso bioinformático crítico.

### Herramientas y Documentación Oficial de Consulta
* **MUSCLE:** [Manual Oficial de MUSCLE (Edgar, Drive5)](https://drive5.com/muscle/manual/)
* **SciPy Spatial Distance:** [Documentación Oficial de SciPy Distance](https://docs.scipy.org/doc/scipy/reference/spatial.distance.html)
* **scikit-learn MDS / PCoA:** [Documentación Oficial de scikit-learn MDS](https://scikit-learn.org/stable/modules/generated/sklearn.manifold.MDS.html)
* **Matplotlib Scatter:** [Documentación Oficial de Matplotlib Pyplot Scatter](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html)

## Punto 5: Señales Evolutivas y Reconstrucción Filogenética

### Contexto Bioinformático
Una vez obtenido el alineamiento múltiple depurado y libre de OFDs, procedemos a inferir la historia evolutiva y las relaciones de parentesco del nuevo aislado de *Halomonas*:

* **Modelo de Sustitución Nucleotídica de Jukes y Cantor (JC69):**
  La distancia cruda de Hamming subestima la verdadera distancia evolutiva entre dos secuencias debido a eventos de **sustituciones múltiples no observadas** en la misma posición a lo largo de millones de años (p. ej. mutaciones reversas $A \to T \to A$ o transversiones consecutivas). El modelo de Jukes-Cantor asume que las cuatro bases ocurren con igual frecuencia ($\pi_A = \pi_C = \pi_G = \pi_T = 0.25$) y que todas las sustituciones ocurren a una tasa constante $\alpha$. La distancia corregida $d_{JC}$ se calcula como:

  $$d_{JC} = -\frac{3}{4} \ln\left(1 - \frac{4}{3} p\right)$$

  donde $p$ es la fracción de sitios disímiles observados entre las dos secuencias.
* **Algoritmo Neighbor-Joining (NJ):**
  Es un método heurístico de reconstrucción filogenética basado en distancias desarrollado por Saitou y Nei (1987). A diferencia de métodos como UPGMA (que asumen una tasa de mutación constante o reloj molecular estricto), **Neighbor-Joining no asume reloj molecular** y permite diferentes velocidades de evolución en diferentes linajes. Funciona emparejando secuencialmente pares de taxones que minimizan la longitud total del árbol (*criterio de mínima evolución*).
* **Módulos BioPython Phylo:**
  BioPython proporciona el módulo `Bio.Phylo.TreeConstruction` con la clase `DistanceCalculator('jukes-cantor')` para calcular matrices corregidas y `DistanceTreeConstructor` para generar árboles mediante `nj()`. La visualización se realiza mediante `Bio.Phylo.draw()`.
* **Identificación de Taxones Hermanos (Sister Taxa):**
  En filogenia, dos taxones son **hermanos** cuando comparten un nodo ancestral común exclusivo más reciente que con cualquier otro organismo del árbol. La identificación del taxón hermano más cercano de nuestro aislado nos revelará cuál es la mejor alternativa biológica para la bioprospección de ectoína.
