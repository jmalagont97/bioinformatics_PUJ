## El gen HLA-B

El gen HLA-B proporciona las instrucciones para la producción de una proteína superficial
de los leucocitos humanos que desempeña un papel fundamental en el sistema inmunitario.
Forma parte del complejo HLA, que ayuda al sistema inmunitario a distinguir proteínas propias
de las producidas por patógenos. Presenta cientos de versiones (alelos), a cada una de las
cuales se le asigna un número particular (como HLA-B27). Tres alelos están presentes entre
las secuencias del archivo, pero su identidad es lo que el taller debe *descubrir*.

## Análisis de componentes principales (PCA)

El PCA reduce el número de dimensiones en grandes conjuntos de datos a componentes
principales que conservan la mayor parte de la información original.

## Análisis de coordenadas principales (PCoA)

PCoA reduce la dimensionalidad mediante una matriz de distancias, proyectándola en un
espacio de menor dimensión. Es la forma *clásica* del MDS.

## Distancia de Hamming

Número de posiciones en las que difieren dos cadenas de la misma longitud. Como las
secuencias del archivo NO están alineadas entre sí con gaps (todas miden la misma longitud),
la suposición de igual longitud se cumple de forma natural.
